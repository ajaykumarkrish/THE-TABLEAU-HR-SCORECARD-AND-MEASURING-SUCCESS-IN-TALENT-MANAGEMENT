```python
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(16, 6))
sns.histplot(df['math score'], kde=True, bins=20)
plt.title('Distribution of Math Scores')
plt.xlabel('Math Score')
plt.ylabel('Frequency')
plt.grid(axis='y', alpha=0.75)
plt.show()
```

```python
import matplotlib.pyplot as plt
plt.hist(df['math score'])
plt.title('Distribution of Math Scores')
plt.xlabel('Math Score')
plt.ylabel('No. of students')
plt.grid(axis='y', alpha=0.8)
plt.show()
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(8,5))
plt.hist(df['math score'],
         bins=10,
         color='orange',
         edgecolor='black',
         linewidth=1.2,
         alpha=0.8)
plt.title('Distribution of Math Scores', fontsize=16, fontweight='bold', fontfamily='serif')
plt.xlabel('Math Score', fontsize=14, fontstyle='italic', fontfamily='serif')
plt.ylabel('Number of Students', fontsize=14, fontstyle='italic', fontfamily='serif')
plt.xticks(fontsize=12, fontfamily='serif')
plt.yticks(fontsize=12, fontfamily='serif')
plt.grid(axis='y', alpha=0.6, linestyle='--')

plt.show()
```

```python
from matplotlib.patches import bbox_artist
df.groupby('test preparation course')[['math score', 'reading score','writing score']].mean().plot(kind='bar', figsize=(5,5))
plt.title('Distribution of scores', fontsize=16, fontweight='bold', fontfamily='serif')
plt.xlabel('Test prepartion score', fontsize=14, fontstyle='italic', fontfamily='serif')
plt.ylabel('Scores', fontsize=14, fontstyle='italic', fontfamily='serif')
plt.legend(title = 'subject', bbox_to_anchor=(1.05, 1))

plt.xticks(fontsize=12, fontfamily='serif', rotation= 0)
plt.yticks(fontsize=12, fontfamily='serif', rotation= 0)
```

```python
df.groupby('parental level of education')[['math score', 'writing score', 'reading score']].mean().reindex(edu_order).plot(kind = 'hist', figsize=(10,5))
plt.title('Average Scores by Parental Level of Education', fontsize=16, fontweight='bold', fontfamily='serif')
plt.xlabel('Parental Level of Education', fontsize=14, fontstyle='italic', fontfamily='serif')
plt.ylabel('Average Score', fontsize=14, fontstyle='italic', fontfamily='serif')
plt.legend(title = 'Subject', bbox_to_anchor=(1.05, 1))
plt.xticks(fontsize=12, fontfamily='serif', rotation=45, ha='right')
plt.yticks(fontsize=12, fontfamily='serif')
plt.tight_layout()
plt.show()
```

```python
edu_order=['some high school','high school','some college',"associate's degree","bachelor's degree","master's degree"]
```
